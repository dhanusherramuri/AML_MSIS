```python
#Matplotlib is used to respresent the data in representative figures and is used to convert the numeric input into figures so as to keep it easy to read



import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,5,6,3,9]

plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
from sklearn.datasets import load_iris
iris = load_iris(as_frame = True)
df = iris.frame
df.head()
```

```python
plt.scatter (df['sepal length (cm)'],
             df['petal length (cm)'])

plt.title ("Scatter Plot")
plt.xlabel ("Sepal Length")
plt.ylabel ("Petal Length")
plt.show()
```

```python
#HISTOGRAM

plt.hist(df['petal width (cm)'])

plt.title("Distribution")
plt.show()
```
