```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/AML_MSIS/titanic.csv")
df.head()
```

```python
df ['Age'] = df ['Age'].fillna(0)
surv = df[df ['Survived'] == 1 ]
non_surv = df[df ['Survived'] == 0 ]


#total number of passengers
print ("\n Total Passengers : " ,len(df))


#average age of survivors and non survivors
print ("\n Average age of Survivors : " ,surv['Age'].mean())
print ("\n Average age of Non Survivors : " ,non_surv['Age'].mean())

#passengers older than 60
print ("\n Passengers older than 60 : ", len(df[df['Age'] > 60]))


#average fare for each passenger pclass
c1 = df[df['Pclass'] == 1]
c2 = df[df['Pclass'] == 2]
c3 = df[df['Pclass'] == 3]
print ("\n Average fare for each passenger pclass 1 : ", c1['Fare'].mean() )
print ("\n Average fare for each passenger pclass 2 : ", c2['Fare'].mean() )
print ("\n Average fare for each passenger pclass 3 : ", c3['Fare'].mean() )
```
